- span simple sentence

```python
text="Python NLTK is powerful. Try it today!"
text
```

```python
spans=list(white_space_tokenizer.span_tokenize(text))
spans
```

```python
for i ,(start,end) in enumerate(spans):
    token=text[start:end]
    print(f"Token {i}: {token} at positions {start}-{end}")
```

- span multiple sentences

```python
sentences={"NLTK makes text processing easy.","WhitespaceTokenizer splits on whitespace.","Perfect for preprocessing tasks."}
sentences
```

```python
for row,sentence in enumerate(sentences):
    tokens=white_space_tokenizer.tokenize(sentence)
    print(f"sentence {row+1}: {tokens}")
```

```python
all_spans=[list(white_space_tokenizer.span_tokenize(sent)) for sent in sentences]
all_spans
```

3) dictionary based 

```python
custom_dict=[("San","Francisco"),("New","York"),("United","Nations")]
custom_dict
```

```python
samlpe_text="San Francisco is beautiful city. The United Nations meets regularly"
cleaned_text=samlpe_text.lower()
cleaned_text
```

```python
tokens=word_tokenize(samlpe_text)
print("Tokenized Text :",tokens)
```

```python
token=MWETokenizer(custom_dict)
tokens_text=token.tokenize(tokens)
print("dictionary based tokenized text :",tokens_text)
```

```python
custom_dict.extend([("machine","learning"),("natrual","language","processing")])
custom_dict
```

```python
sentences=[
    "San Francisco is a beautiful place.",
    "The United Nations is headquartered in New York.",
    "Machine Learning is a subset of Artificial Intelligence."
]
```

```python
tokenizer=MWETokenizer(custom_dict)
```

```python
for sentence in sentences:
    cleaned_text = sentence.lower()
    tokens = word_tokenize(cleaned_text)
    print(tokenizer.tokenize(tokens))
```
